```python
#real world data
import numpy as np
import seaborn as sns
import pandas as pd
import random

titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
null_values = titanic.isnull().sum()
for index, value in null_values.items():
  if value > 0:
    print(index, value)
```

```python
#Approaches to Handle Missing Values

# 1. Drop missing values
titanic_dropped = titanic.dropna()  # Dropping rows with any missing values
titanic_dropped.shape

# 2. Impute Missing Values
titanic['age'].fillna(titanic['age'].mean(), inplace=True)  # Mean/Median Imputation for Age
titanic['embarked'].fillna(titanic['embarked'].mode()[0], inplace=True)  # Mode Imputation for Embarked
titanic['age'].fillna(method='ffill', inplace=True)  # Forward Fill for Age
titanic['age'].fillna(method='bfill', inplace=True)  # Backward Fill for Age
titanic['embark_town'].fillna(random.choice(titanic['embark_town'].dropna().unique()), inplace=True)  # Random Imputation for Embark Town

null_values = titanic.isnull().sum()
for index, value in null_values.items():
  if value > 0:
    print(index, value)
```

```python
# 3. Predictive Modeling for Age

titanic = sns.load_dataset('titanic')
titanic.head()

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# Separate out data with missing ages BEFORE filling in
titanic_missing_age = titanic[titanic['age'].isna()]
titanic_no_missing_age = titanic.dropna(subset=['age'])

numeric_features = titanic_no_missing_age.select_dtypes(include=[np.number]).drop(columns=['age'])
X_train, X_test, y_train, y_test = train_test_split(numeric_features, titanic_no_missing_age['age'], test_size=0.2)

model = LinearRegression()
model.fit(X_train, y_train)

# Predict missing ages ONLY on the portion of the data that was missing ages
# Use the SAME columns for prediction as were used in training
titanic.loc[titanic['age'].isna(), 'age'] = model.predict(titanic_missing_age.select_dtypes(include=[np.number]).drop(columns=['age']))

null_values = titanic.isnull().sum()
for index, value in null_values.items():
  if value > 0:
    print(index, value)
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Step 1: Generate the raw data
np.random.seed(42)  # For reproducibility

# Generate 400 random heights in feet (range from 5.0 to 7.0)
heights = np.random.uniform(5.0, 7.0, 400)

# Generate 400 random weights in grams (range from 50,000 to 150,000)
weights = np.random.uniform(50000, 150000, 400)

# DataFrame
data = pd.DataFrame({
    'Height (feet)': heights,
    'Weight (grams)': weights
})

# Step 2: Apply MinMax Normalization to the weights
min_max_scaler = MinMaxScaler()
data['Weight (MinMax)'] = min_max_scaler.fit_transform(data[['Weight (grams)']])

# Step 3: Apply Z-score Normalization to the weights
z_score_scaler = StandardScaler()
data['Weight (Z-score)'] = z_score_scaler.fit_transform(data[['Weight (grams)']])

# Display the result
data.head()
```
