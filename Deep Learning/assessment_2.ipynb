```python
# brief about multilayer perceptron

# A multilayer perceptron (MLP) is a class of feedforward artificial neural network (ANN).
# An MLP consists of at least three layers of nodes: an input layer, a hidden layer and an output layer.
# Except for the input nodes, each node is a neuron that uses a nonlinear activation function.
# MLP utilizes a supervised learning technique called backpropagation for training.
# MLPs are useful in research for their ability to solve problems stochastically.
```

```python
# brief about BPN

# Backpropagation (BPN) is the central mechanism by which artificial neural networks learn.
# It is the messenger telling the network whether or not it made a mistake when it made a prediction.
# A BPN is a supervised learning algorithm, for training multi-layer perceptrons (MLPs).
```

```python
# brief about mini-batch gradient descent

# Mini-batch gradient descent is a variation of the gradient descent optimization algorithm commonly used in training machine learning models, particularly neural networks.
# Mini-batch gradient descent addresses issue by dividing the training dataset into smaller, randomly selected batches (mini-batches).
# The gradient is then calculated and the parameters are updated for each mini-batch.
# This process is repeated for a specified number of epochs.
# Common mini-batch sizes range from tens to hundreds of samples.
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import SGD

# MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Flatten images and normalize pixel values
x_train = x_train.reshape(60000, 784).astype('float32') / 255
x_test = x_test.reshape(10000, 784).astype('float32') / 255

# Convert labels to categorical
y_train = keras.utils.to_categorical(y_train, num_classes=10)
y_test = keras.utils.to_categorical(y_test, num_classes=10)

# MLP model
model = Sequential()
model.add(Dense(256, activation='relu', input_shape=(784,)))
model.add(Dense(128, activation='relu'))
model.add(Dense(10, activation='softmax'))

# categorical crossentropy loss
optimizer = SGD(learning_rate=0.01)
model.compile(loss='categorical_crossentropy', optimizer=optimizer, metrics=['accuracy'])

# mini-batch gradient descent
batch_size = 64
epochs = 25
model.fit(x_train, y_train, batch_size=batch_size, epochs=epochs, validation_data=(x_test, y_test))

# Evaluate the mode
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
print('Test loss:', loss)
print('Test accuracy:', accuracy)
```
